## 1. Motivation: Unifying Discrete Optimization, Sequential Control, and Partial Observability

Throughout this course, we have explored five distinct mathematical disciplines that are traditionally taught in disparate university departments:
1. **Submodular Optimization** (Theoretical Computer Science & Combinatorics)
2. **Markov Decision Processes & Dynamic Programming** (Operations Research & Reinforcement Learning)
3. **Partially Observable MDPs & Belief Space Control** (Robotics & Artificial Intelligence)
4. **Multi-Armed Bandits & Restless Index Policies** (Applied Probability & Economics)
5. **Sequential Analysis & Quickest Change Detection** (Mathematical Statistics & Signal Processing)

In industrial systems engineering, however, real-world problems do not respect academic departmental boundaries:
* A test selection pipeline requires **submodular maximization** to prune redundant coverage, while managing **POMDP belief states** over flaky test health.
* Allocating soak-test runners across degraded hardware targets requires **restless Whittle indexation** to eliminate quarantine blindness.
* Switching a production microservice between primary and fallback clusters requires **$(s, S)$ hysteresis impulse control** to prevent chattering under load noise.
* Catching newly merged regressions requires **Page's CUSUM quickest change detection** to minimize detection latency under Lorden's bound.

This Appendix provides the **Master Rosetta Stone**: a unified mathematical dictionary, a diagnostic decision tree for practitioners, and an exhaustive formula cheat sheet.

## 3. Practitioner's Diagnostic Decision Tree

When confronted with a complex infrastructure design or optimization problem, use this systematic diagnostic protocol:

```
                                  [Problem Encountered]
                                            |
                         Is the decision dynamic over time?
                                   /                 \
                             [NO]                    [YES]
                              /                         \
                 Static Portfolio Selection       Is state fully observable?
                 ==========================              /             \
                  Do returns exhibit                    [YES]          [NO]
                  diminishing marginal value?            /                \
                       /              \            Standard MDP      Does observation
                     [YES]            [NO]         ============      depend on action?
                      /                \           (Bellman DP)        /          \
              Submodular Max      Linear/Integer                     [YES]        [NO]
              ==============       Programming                         /             \
              (Nemhauser Bound)   (Simplex/Branch&Bound)           POMDP /        Quickest Change
                                                                 Restless Bandit    Detection
                                                                 ===============  ===============
                                                                 (Whittle Index)   (Page's CUSUM)
```

### Diagnostic Checklist
1. **Diminishing Returns (Submodularity)**: Does adding an item (e.g., a test, a cache node, an alert rule) provide less incremental benefit when added to a large existing set than when added to a small set? If yes, formulate as **Submodular Optimization**.
2. **Dynamic Transitions & Full State**: Can the exact state of the system (e.g., queue size, available memory) be measured without noise, and do actions transition the state? If yes, formulate as a **Markov Decision Process (MDP)**.
3. **Hidden State & Action-Dependent Information**: Is the true health (clean vs faulty) latent, and does testing or probing the entity cost resources and reveal information? If yes, formulate as a **Partially Observable MDP (POMDP)**.
4. **Multiple Competing Restless Entities**: Are there $K$ entities competing for $M < K$ service slots, where unserved entities continue to evolve (e.g., healing or degrading while idle)? If yes, formulate as a **Restless Multi-Armed Bandit (Whittle Index)**.
5. **Streaming Drift / Anomaly Detection**: Is an entity streaming measurements continuously, and you must detect an unannounced shift in distribution as fast as possible with bounded false alarms? If yes, formulate as **Quickest Change Detection (CUSUM / Shiryaev-Roberts)**.

## 4. Complete Mathematical Formula Cheat Sheet

### 4.1 Submodular Optimization
* **Definition of Submodularity**: For ground set $\Omega$ and set function $F: 2^\Omega \to \mathbb{R}$:

$$
F(A \cup \{e\}) - F(A) \ge F(B \cup \{e\}) - F(B), \quad \forall A \subseteq B \subseteq \Omega, \; e \in \Omega \setminus B
$$

* **Nemhauser-Wolsey Theorem (1978)**: For a monotone submodular function $F$ with $F(\emptyset) = 0$ under cardinality constraint $|S| \le k$, the greedy algorithm achieves:

$$
F(S_{\text{greedy}}) \ge \left(1 - \frac{1}{e}\right) \max_{|S| \le k} F(S) \approx 0.6321 \cdot F(S^*)
$$

* **Lazy Greedy Acceleration (Minoux 1978)**: Maintains a max-heap of marginal gains $\Delta(e) = F(S \cup \{e\}) - F(S)$. By submodularity, $\Delta(e)$ can only decrease, enabling massive pruning of function evaluations ($> 90\%$ speedup).

### 4.2 Partially Observable Markov Decision Processes (POMDP)
* **Belief State Update**: For prior belief $b(s)$, action $a$, and observation $y$:

$$
b'(s') = \frac{O(y \mid s', a) \sum_{s \in \mathcal{S}} P(s' \mid s, a) b(s)}{\mathbb{P}(y \mid b, a)}
$$

* **Value Function Convexity (Sondik 1971)**: The optimal finite-horizon value function $V_t(b)$ is Piecewise Linear and Convex (PWLC):

$$
V_t(b) = \max_{\alpha \in \Gamma_t} \sum_{s \in \mathcal{S}} \alpha(s) b(s) = \max_{\alpha \in \Gamma_t} \langle \alpha, b \rangle
$$

### 4.3 Restless Bandits & Whittle Index
* **Lagrangian Relaxation**: Decouples $K$-arm budget $\sum a_{k, t} \le M$ via passivity subsidy $\lambda$:

$$
V(s; \lambda) = \max \left\{ R(s, 1) + \gamma \sum_{s'} P^{(1)}(s' \mid s) V(s'; \lambda), \; R(s, 0) + \lambda + \gamma \sum_{s'} P^{(0)}(s' \mid s) V(s'; \lambda) \right\}
$$

* **Whittle Index Definition**:

$$
W(s) = \inf \{ \lambda \in \mathbb{R} : V(s, 0; \lambda) = V(s, 1; \lambda) \}
$$

### 4.4 Quickest Change Detection & Optimal Stopping
* **Wald SPRT Decision Boundaries**:

$$
a = \ln \left(\frac{1 - \beta}{\alpha}\right), \quad b = \ln \left(\frac{\beta}{1 - \alpha}\right)
$$

* **Page's CUSUM Recursive Update**:

$$
S_t = \max\left(0, \; S_{t-1} + \ln \frac{p_1(y_t)}{p_0(y_t)}\right), \quad T_{\text{CUSUM}} = \inf \{t : S_t \ge h\}
$$

* **Lorden's Minimax Delay Lower Bound**:

$$
\bar{\mathbb{E}}_1[T - \tau] \ge \frac{\ln(\text{ARL}_0)}{D_{\text{KL}}(P_1 \parallel P_0)} \quad \text{as } \text{ARL}_0 \to \infty
$$

## 5. Concrete Worked Problem Mappings with Real Numbers

### 5.1 Case 1: Submodular Feature / Sensor Selection
* **Scenario**: A monitoring engineer has 10 candidate synthetic latency probes. Each probe covers a subset of 100 backend database shards.
* **Problem**: Select $k = 3$ probes to maximize unique shard coverage.
* **Coverage Matrix**:
  * Probe 1: covers shards $\{1..40\}$ ($F = 40$)
  * Probe 2: covers shards $\{30..70\}$ ($F = 41$)
  * Probe 3: covers shards $\{60..90\}$ ($F = 31$)
  * Probe 4: covers shards $\{1..30\}$ ($F = 30$)
* **Greedy Step 1**: Probe 2 chosen ($|S_1| = 41$).
* **Greedy Step 2**: Marginal gains:
  * Probe 1: shards $\{1..29\} \implies \Delta = 29$.
  * Probe 3: shards $\{71..90\} \implies \Delta = 20$.
  * Probe 1 chosen ($|S_2| = 41 + 29 = 70$).
* **Greedy Step 3**: Marginal gains:
  * Probe 3: shards $\{71..90\} \implies \Delta = 20$.
  * Probe 3 chosen ($|S_3| = 70 + 20 = 90$).
* Total coverage $= 90/100$ ($90\%$). Optimal coverage is $90\%$. Nemhauser bound guaranteed at least $1 - 1/e \approx 63.2\%$.

### 5.2 Case 2: CUSUM Drift Detection on CI Runner Failures
* Pre-change noise: $p_0 = 0.005$ ($0.5\%$ error).
* Post-change degradation: $p_1 = 0.04$ ($4.0\%$ error).
* False alarm target: $\text{ARL}_0 = 50{,}000$ runs ($h = \ln(50000) \approx 10.82$ nats).
* KL Divergence:
  $$D_{\text{KL}}(0.04 \parallel 0.005) = 0.04 \ln(8) + 0.96 \ln(0.96 / 0.995) = 0.04(2.0794) + 0.96(-0.0358) = 0.08318 - 0.03437 = 0.04881 \text{ nats}$$
* Lorden's Expected Detection Delay:
  $$\mathbb{E}[\text{Delay}] \approx \frac{10.82}{0.04881} \approx 221.7 \text{ runs}$$

## 6. Synthesis and Course Conclusions

The overarching insight of this curriculum is that **statistical uncertainty and computational scale are duals of one another**:
1. When observations are deterministic but combinations are massive $\to$ **Submodular Optimization** provides the polynomial-time approximation guarantee.
2. When combinations are dynamic and states are stochastic $\to$ **Markov Decision Processes** and **Bellman equations** guarantee global optimality.
3. When states are hidden and measurements are noisy $\to$ **POMDPs**, **Belief Spaces**, and **$\alpha$-vectors** transform uncertainty into geometry.
4. When multiple unobserved entities evolve autonomously $\to$ **Whittle Restless Indices** conquer the PSPACE curse of dimensionality.
5. When streaming systems drift without warning $\to$ **Wald SPRT** and **Page CUSUM** provably minimize the lag of discovery.
